```python
#@title Setup: imports, mount, base paths
import pandas as pd
from pathlib import Path

# Mount Drive if in Colab
try:
    from google.colab import drive
    drive.mount('/content/drive')
    BASE_DIR = Path('/content/drive/MyDrive/ai4trade')
except Exception:
    # If running locally / JupyterLab
    BASE_DIR = Path.home() / 'ai4trade'

INTERIM = BASE_DIR / 'data' / 'interim'

# Quick peek
print("Working directory:", BASE_DIR)
print("Interim data dir  :", INTERIM)
```

```python
#@title Utilities: load partner set & filter/safe-write

def load_partner_set(parquet_path: Path) -> set:
    """
    Load a parquet file and return the set of unique 'destination' partners present.
    """
    df = pd.read_parquet(parquet_path, columns=['destination'])
    partners = set(df['destination'].dropna().astype(str).unique())
    print(f"Loaded {parquet_path.name}: {len(partners)} partners detected.")
    return partners

def filter_by_partners_and_save(src_path: Path, partners: set, suffix: str = "_key_partners"):
    """
    Read src parquet, filter to destination ∈ partners, write alongside with suffix.
    Derive hs4 from hs6, ensuring it's a string to preserve leading zeros.
    Returns output path.
    """
    df = pd.read_parquet(src_path)
    before = len(df)

    # Derive hs4 from hs6 as string to preserve leading zeros
    df['hs4'] = df['hs6'].astype(str).str[:4]

    df_f = df[df['destination'].astype(str).isin(partners)].copy()
    after = len(df_f)
    out_path = src_path.with_name(src_path.stem + f"{suffix}.parquet")
    df_f.to_parquet(out_path, index=False)
    print(f"Saved {out_path.name}: {after:,} rows (from {before:,})")
    return out_path
```

```python
#@title Get partner set from CHN_imports_2025.parquet
chn_imp_2025 = INTERIM / "CHN_imports_2025.parquet"
partners_CHN_Imp_2025 = load_partner_set(chn_imp_2025)
```

```python
#@title Filter CHN_imports_2023 to 2025 partners and save
src = INTERIM / "CHN_imports_2023.parquet"
out_CHN_imp_2023 = filter_by_partners_and_save(src, partners_CHN_Imp_2025)
```
